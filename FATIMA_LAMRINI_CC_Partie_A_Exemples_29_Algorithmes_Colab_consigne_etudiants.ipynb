```python
# === Outils communs pour les schémas (à exécuter UNE SEULE FOIS, avant l'algorithme 1) ===
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch, Circle, Ellipse
plt.rcParams.update({'font.size': 10, 'axes.titlesize': 11})

def boite(ax, x, y, texte, couleur='#dbe9f6', l=2.6, h=1.0, taille=9):
    ax.add_patch(FancyBboxPatch((x-l/2, y-h/2), l, h, boxstyle='round,pad=0.05', fc=couleur, ec='#33415c', lw=1.2))
    ax.text(x, y, texte, ha='center', va='center', fontsize=taille)

def fleche(ax, p1, p2, texte='', rad=0.0, style='-', dy=0.25):
    ax.add_patch(FancyArrowPatch(p1, p2, arrowstyle='-|>', mutation_scale=14, lw=1.4, ls=style,
                                 color='#33415c', connectionstyle=f'arc3,rad={rad}'))
    if texte:
        ax.text((p1[0]+p2[0])/2, (p1[1]+p2[1])/2+dy, texte, ha='center', fontsize=8, color='#7a1f1f')

def flux(etapes, titre, boucle=None, taille=9):
    """Diagramme de flux horizontal. boucle=(i, j, 'texte') : flèche de retour de l'étape i vers l'étape j."""
    n = len(etapes)
    fig, ax = plt.subplots(figsize=(2.9*n, 3.0))
    for i, t in enumerate(etapes):
        boite(ax, 3.2*i, 0, t, l=2.7, h=1.4, taille=taille, couleur='#dbe9f6' if i < n-1 else '#e2f0d9')
        if i < n-1:
            fleche(ax, (3.2*i+1.4, 0), (3.2*(i+1)-1.4, 0))
    if boucle:
        i, j, txt = boucle
        fleche(ax, (3.2*i, 0.8), (3.2*j, 0.8), rad=0.35)
        ax.text(3.2*(i+j)/2, 1.55, txt, ha='center', fontsize=8, color='#7a1f1f')
    ax.set_xlim(-1.6, 3.2*(n-1)+1.6); ax.set_ylim(-1.1, 2.0); ax.axis('off')
    ax.set_title(titre); plt.tight_layout(); plt.show()
```

### Q1

- **Nom, famille, objectif :** Régression linéaire (*Linear Regression*) — famille **régression** (supervisé). Objectif : prédire une variable cible continue à partir de variables explicatives.
- **Intuition et étapes :** on cherche l'hyperplan qui passe au plus près des points. Entrée : matrice $X$ et cible $y$ d'entraînement → apprentissage : estimation des coefficients par moindres carrés → sortie : prédiction $\hat y$ pour de nouvelles observations.
- **Formule :** $\hat y=\beta_0+\sum_{j=1}^{p}\beta_j x_j$ avec $\hat\beta=\arg\min_\beta\sum_{i=1}^{n}(y_i-\hat y_i)^2$ ($n$ : nombre d'observations, $p$ : nombre de variables, $\beta_0$ : ordonnée à l'origine, $\beta_j$ : effet de la variable $x_j$).
- **Paramètres / hyperparamètres :** paramètres appris : $\beta_0,\dots,\beta_p$ ; hyperparamètres : aucun essentiel (seulement le choix d'inclure ou non l'ordonnée à l'origine).
- **Avantages :** (1) très interprétable (chaque $\beta_j$ a un sens) ; (2) rapide à entraîner et sans réglage.
- **Limites :** (1) suppose une relation linéaire ; (2) sensible aux valeurs extrêmes et à la colinéarité entre variables.

### Q2. Schéma explicatif

**Comment le lire :** Les points sont les observations. La droite rouge est la prédiction du modèle. Chaque trait pointillé gris est un résidu (écart entre la valeur réelle et la valeur prédite) : le modèle retient la droite qui minimise la somme des carrés de ces écarts.

```python
rng = np.random.default_rng(42)
x = np.sort(rng.uniform(0, 10, 25)); y = 2*x + 3 + rng.normal(0, 3, 25)
b1, b0 = np.polyfit(x, y, 1); yp = b0 + b1*x
plt.figure(figsize=(6, 4))
plt.scatter(x, y, label=r'observations $(x_i, y_i)$')
plt.plot(x, yp, 'r', label=r'droite ajustée $\hat{y}$')
plt.vlines(x, y, yp, colors='gray', linestyles=':', label=r'résidus $y_i-\hat{y}_i$')
plt.xlabel('variable explicative $x$'); plt.ylabel('cible $y$')
plt.title('Régression linéaire : minimiser la somme des carrés des résidus')
plt.legend(); plt.show()
```

### Q3. Contextes d'utilisation

- **Deux cas d'usage concrets :**
  1. *Marketing / gestion commerciale :* prédire les ventes mensuelles d'un produit à partir du budget publicitaire (télévision, web, radio), du prix et du nombre de points de vente. Les coefficients indiquent l'effet moyen d'une unité de budget supplémentaire sur les ventes.
  2. *Logistique / commerce international :* estimer le coût d'expédition d'une marchandise à partir de la distance, du poids et du volume du colis. Le modèle sert à établir un devis ou à budgéter des flux de transport.
- **Nature des données et prérequis techniques :** une cible **numérique continue** et des variables explicatives numériques (les variables catégorielles doivent être encodées, par exemple en indicatrices 0/1). Il faut plus d'observations que de variables, des données sans valeurs manquantes (ou imputées), peu de valeurs extrêmes et une relation approximativement **linéaire** entre variables et cible. Les hypothèses classiques portent sur les résidus : moyenne nulle, variance constante, indépendance entre observations, et absence de forte colinéarité entre variables explicatives. La standardisation n'est pas indispensable à l'ajustement, mais elle facilite la comparaison des coefficients entre variables d'échelles différentes.
- **Cas peu approprié :** prédire si un client va faire **défaut de paiement** (cible binaire 0/1). La régression linéaire peut produire des valeurs inférieures à 0 ou supérieures à 1, qui ne sont pas des probabilités, et suppose des erreurs de variance constante, ce qui n'a pas de sens pour une cible binaire. On préfère alors la régression logistique. Elle est aussi peu adaptée quand la relation est nettement non linéaire (saturation, effet de seuil) ou quand les observations sont des séries temporelles autocorrélées.
- **Métriques adaptées :** **MAE** (erreur absolue moyenne, dans l'unité de la cible), **RMSE** (racine de l'erreur quadratique moyenne, pénalise davantage les grosses erreurs) et **R²** (part de variance de la cible expliquée par le modèle). Elles se calculent sur des données de test non utilisées pour l'apprentissage. On complète par l'analyse graphique des **résidus** (valeurs prédites en abscisse, résidus en ordonnée) pour vérifier les hypothèses.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

- **Nom, famille, objectif :** Régression Ridge (*Ridge Regression*) — famille **régression**. Objectif : prédire une variable continue en limitant l'instabilité des coefficients (régularisation L2).
- **Intuition et étapes :** même modèle que la régression linéaire, mais on pénalise les coefficients trop grands. Entrée : $X$ standardisée et $y$ → minimisation de l'erreur + pénalité → prédiction $\hat y$.
- **Formule :** $\min_\beta\sum_{i=1}^{n}(y_i-\hat y_i)^2+\alpha\sum_{j=1}^{p}\beta_j^2$ ($\alpha\ge 0$ : force de la pénalité ; $\alpha=0$ redonne les moindres carrés).
- **Paramètres / hyperparamètres :** paramètres : les $\beta_j$ ; hyperparamètre : $\alpha$ (à choisir par validation, jamais sur le test final).
- **Avantages :** (1) stabilise les coefficients en présence de variables corrélées ; (2) réduit la variance donc le surapprentissage.
- **Limites :** (1) ne met aucun coefficient exactement à zéro (pas de sélection de variables) ; (2) nécessite de standardiser les variables et introduit un biais.

### Q2. Schéma explicatif

**Comment le lire :** Chaque courbe est un coefficient $\beta_j$. À gauche ($\alpha$ petit) on retrouve les moindres carrés ; en allant vers la droite ($\alpha$ grand), tous les coefficients se rapprochent de 0 progressivement, mais sans jamais s'annuler exactement.

```python
from sklearn.linear_model import Ridge
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=100, n_features=6, n_informative=3, noise=10, random_state=42)
alphas = np.logspace(-2, 4, 40)
coefs = [Ridge(alpha=a).fit(X, y).coef_ for a in alphas]
plt.figure(figsize=(6, 4)); plt.plot(alphas, coefs); plt.xscale('log'); plt.axhline(0, color='k', lw=.5)
plt.legend([rf'$\beta_{j+1}$' for j in range(6)], ncol=3, fontsize=8)
plt.xlabel(r'$\alpha$ (échelle logarithmique)'); plt.ylabel(r'valeur du coefficient $\beta_j$')
plt.title(r'Ridge : les coefficients rétrécissent quand $\alpha$ augmente'); plt.show()
```

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Présentation

- **Nom, famille, objectif :** Lasso (*Least Absolute Shrinkage and Selection Operator*) — famille **régression**. Objectif : prédire une variable continue tout en sélectionnant automatiquement les variables utiles (régularisation L1).
- **Intuition et étapes :** la pénalité en valeur absolue force certains coefficients à valoir exactement zéro. Entrée : $X$ standardisée et $y$ → minimisation du critère pénalisé → modèle parcimonieux et prédiction.
- **Formule :** $\min_\beta\ \frac{1}{2n}\sum_{i=1}^{n}(y_i-\hat y_i)^2+\alpha\sum_{j=1}^{p}|\beta_j|$ ($\alpha>0$ : force de la pénalité).
- **Paramètres / hyperparamètres :** paramètres : les $\beta_j$ (dont beaucoup égaux à 0) ; hyperparamètre : $\alpha$.
- **Avantages :** (1) sélection de variables intégrée ; (2) modèle parcimonieux, donc plus lisible.
- **Limites :** (1) parmi des variables fortement corrélées, en retient une de façon arbitraire et instable ; (2) pas de solution fermée (optimisation itérative) et standardisation nécessaire.

### Q2. Schéma explicatif

**Comment le lire :** Chaque courbe est un coefficient $\beta_j$. Quand $\alpha$ augmente, certains coefficients tombent exactement à 0 (la variable est éliminée) : c'est la sélection de variables propre au Lasso, contrairement à Ridge.

```python
from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=100, n_features=6, n_informative=3, noise=10, random_state=42)
alphas = np.logspace(-2, 1.5, 40)
coefs = [Lasso(alpha=a, max_iter=10000).fit(X, y).coef_ for a in alphas]
plt.figure(figsize=(6, 4)); plt.plot(alphas, coefs); plt.xscale('log'); plt.axhline(0, color='k', lw=.5)
plt.legend([rf'$\beta_{j+1}$' for j in range(6)], ncol=3, fontsize=8)
plt.xlabel(r'$\alpha$ (échelle logarithmique)'); plt.ylabel(r'valeur du coefficient $\beta_j$')
plt.title(r'Lasso : des coefficients deviennent exactement nuls'); plt.show()
```

- **Limites :** (1) surapprentissage rapide quand $d$ augmente ; (2) le nombre de variables explose avec $p$ et $d$, et l'extrapolation hors du domaine observé est peu fiable.

### Présentation

- **Nom, famille, objectif :** Arbre de régression (*Regression Tree*) — famille **régression**. Objectif : prédire une valeur continue par des règles « si… alors… » successives.
- **Intuition et étapes :** on découpe récursivement l'espace des variables en zones homogènes. Entrée : $X,y$ → à chaque nœud, choix de la variable et du seuil qui réduisent le plus l'erreur → arrêt (profondeur, taille de feuille) → prédiction = moyenne de $y$ dans la feuille atteinte.
- **Formule :** on choisit $(j,s)$ minimisant $\sum_{x_i\in R_1}(y_i-\bar y_{R_1})^2+\sum_{x_i\in R_2}(y_i-\bar y_{R_2})^2$ ($j$ : variable, $s$ : seuil, $R_1,R_2$ : les deux zones obtenues, $\bar y_R$ : moyenne de $y$ dans $R$).
- **Paramètres / hyperparamètres :** paramètres : variables et seuils de coupure, valeurs des feuilles ; hyperparamètres : profondeur maximale, taille minimale des feuilles.
- **Avantages :** (1) très lisible (règles explicites) ; (2) capte des non-linéarités et interactions sans standardisation.
- **Limites :** (1) instable et prompt au surapprentissage si trop profond ; (2) prédictions constantes par morceaux, pas d'extrapolation.

### Présentation

- **Nom, famille, objectif :** Gradient Boosting — régression (*Gradient Boosting Regressor*) — famille **régression** (ensemble séquentiel). Objectif : obtenir une prédiction continue très précise sur données tabulaires. L'exemple utilise la variante par histogrammes de scikit-learn.

- **Avantages :** (1) souvent parmi les plus précis sur données tabulaires ; (2) perte et variables de types variés gérées facilement.
- **Limites :** (1) sensible au réglage et au surapprentissage si trop d'itérations ; (2) entraînement séquentiel et modèle peu interprétable.

### Présentation

- **Nom, famille, objectif :** Régression logistique (*Logistic Regression*) — famille **classification** (supervisé). Objectif : estimer la probabilité d'appartenance à une classe (ici binaire) puis décider.
- **Intuition et étapes :** un score linéaire est transformé en probabilité par la fonction sigmoïde. Entrée : $X$ standardisée et étiquettes $y\in\{0,1\}$ → apprentissage des coefficients par minimisation de la log-vraisemblance négative → probabilité puis classe (seuil, souvent 0,5).

- **Paramètres / hyperparamètres :** paramètres : $\beta_0,\beta$ ; hyperparamètres : force de régularisation (paramètre $C$, inverse de la pénalité), type de pénalité, seuil de décision.
- **Avantages :** (1) fournit des probabilités et des coefficients interprétables (rapports de cotes) ; (2) rapide et robuste avec régularisation.
- **Limites :** (1) frontière de décision linéaire ; (2) sensible aux variables non standardisées et fortement corrélées.

```python
from sklearn.linear_model import LogisticRegression
rng = np.random.default_rng(42)
x = np.r_[rng.normal(-1, 1, 40), rng.normal(1.5, 1, 40)]; y = np.r_[np.zeros(40), np.ones(40)]
m = LogisticRegression().fit(x[:, None], y); g = np.linspace(-4, 5, 200)
plt.figure(figsize=(6, 4)); plt.scatter(x, y, c=y, cmap='coolwarm', edgecolor='k', label='observations (classe 0 / 1)')
plt.plot(g, m.predict_proba(g[:, None])[:, 1], 'k', lw=2, label=r'$P(y=1\mid x)$ (sigmoïde)')
plt.axhline(.5, ls='--', color='gray', label='seuil de décision 0,5')
plt.xlabel('variable $x$'); plt.ylabel('probabilité'); plt.title('Régression logistique'); plt.legend(); plt.show()
```

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

- **Nom, famille, objectif :** K plus proches voisins (*k-Nearest Neighbors, KNN*) — famille **classification** (supervisé, non paramétrique). Objectif : classer une observation selon la classe majoritaire de ses voisins.
- **Intuition et étapes :** « dis-moi qui sont tes voisins, je te dirai qui tu es ». Entrée : $X$ standardisée et $y$ (mémorisés) → aucun apprentissage explicite → pour un nouveau point, calcul des distances à tous les points d'entraînement, sélection des $k$ plus proches, vote majoritaire.
- **Formule :** distance euclidienne $d(x,x')=\sqrt{\sum_{j=1}^{p}(x_j-x'_j)^2}$ ; classe prédite $=$ classe la plus fréquente parmi les $k$ voisins de $x$.
- **Paramètres / hyperparamètres :** pas de paramètres appris (les données servent de modèle) ; hyperparamètres : $k$, distance utilisée, pondération des voisins.
- **Avantages :** (1) très simple à comprendre, aucun entraînement ; (2) frontières de décision non linéaires.
- **Limites :** (1) prédiction lente sur grands jeux de données ; (2) sensible à l'échelle des variables et à la dimension élevée (malédiction de la dimension).

### Q2. Schéma explicatif

**Comment le lire :** Le point noir (étoile) est l'observation à classer. Le cercle entoure ses $k=5$ plus proches voisins (entourés en noir) : la classe majoritaire parmi eux est la classe prédite.

```python
from sklearn.neighbors import NearestNeighbors
rng = np.random.default_rng(42)
A = rng.normal([0, 0], .9, (30, 2)); B = rng.normal([2.5, 2], .9, (30, 2)); X = np.vstack([A, B]); cl = np.r_[np.zeros(30), np.ones(30)]
nouveau = np.array([[1.3, 1.0]])
idx = NearestNeighbors(n_neighbors=5).fit(X).kneighbors(nouveau)[1][0]
rayon = np.linalg.norm(X[idx[-1]] - nouveau[0])
fig, ax = plt.subplots(figsize=(5.5, 5))
ax.scatter(*A.T, c='tab:blue', label='classe A'); ax.scatter(*B.T, c='tab:orange', label='classe B')
ax.scatter(*X[idx].T, s=170, facecolors='none', edgecolors='k', lw=1.5, label='5 plus proches voisins')
ax.scatter(*nouveau.T, marker='*', s=250, c='k', label='point à classer')
ax.add_patch(Circle(nouveau[0], rayon, fill=False, ls='--'))
ax.set_aspect('equal'); ax.set_title(f'KNN (k=5) : vote A={int((cl[idx]==0).sum())}, B={int((cl[idx]==1).sum())}'); ax.legend(); plt.show()
```

### Contextes d'utilisation

**Cas 1 – Finance :** Classer les clients d'une banque selon leur profil financier afin d'identifier ceux qui présentent des caractéristiques similaires.

**Cas 2 – Marketing :** Recommander des produits à des clients en fonction des préférences de consommateurs ayant des profils proches.

**Nature des données et prérequis :** Données structurées numériques, avec des variables comparables. La standardisation est importante car l'algorithme repose sur le calcul des distances. Le choix du nombre de voisins doit être adapté.

**Cas inapproprié :** Il est peu adapté aux très grands jeux de données comportant un grand nombre de variables, car les calculs de distance peuvent devenir coûteux et moins pertinents.

**Métriques :** Accuracy, précision, rappel et F1-score.

### Analyse de l'expérience Python

Le paramètre `n_neighbors=5` signifie que la classe est déterminée à partir des cinq voisins les plus proches.

**Résultat obtenu : F1-score = 0,8889.**

Ce résultat indique une bonne capacité de classification sur l'échantillon de test. Cependant, les performances dépendent du choix du nombre de voisins et de la qualité des données.

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

- **Nom, famille, objectif :** Naive Bayes gaussien (*Gaussian Naive Bayes*) — famille **classification** (probabiliste). Objectif : classer en appliquant le théorème de Bayes.
- **Intuition et étapes :** on suppose les variables indépendantes entre elles conditionnellement à la classe. Entrée : $X,y$ → estimation, pour chaque classe, de sa fréquence et de la moyenne/variance de chaque variable → calcul de la probabilité a posteriori → classe la plus probable.
- **Formule :** $P(c\mid x)\propto P(c)\prod_{j=1}^{p}\mathcal N(x_j\mid\mu_{jc},\sigma_{jc}^2)$ ($c$ : classe, $P(c)$ : probabilité a priori, $\mu_{jc},\sigma^2_{jc}$ : moyenne et variance de la variable $j$ dans la classe $c$).
- **Paramètres / hyperparamètres :** paramètres : $P(c),\mu_{jc},\sigma^2_{jc}$ ; hyperparamètre : lissage de variance (*var_smoothing*).
- **Avantages :** (1) extrêmement rapide, fonctionne avec peu de données ; (2) simple et peu sujet au surapprentissage.
- **Limites :** (1) l'hypothèse d'indépendance est rarement vraie ; (2) probabilités souvent mal calibrées.

### Q2. Schéma explicatif

**Comment le lire :** Chaque courbe est la densité gaussienne d'une variable dans une classe. Pour la valeur observée (trait noir vertical), on compare les hauteurs des deux courbes, pondérées par les probabilités a priori : la classe dont la courbe est la plus haute est prédite.

```python
from scipy.stats import norm
g = np.linspace(-5, 8, 300); x0 = 1.5
plt.figure(figsize=(6, 4))
plt.plot(g, norm.pdf(g, 0, 1.2), 'tab:blue', label=r'classe A : $\mathcal{N}(\mu_A,\sigma_A^2)$')
plt.plot(g, norm.pdf(g, 3, 1.5), 'tab:orange', label=r'classe B : $\mathcal{N}(\mu_B,\sigma_B^2)$')
plt.axvline(x0, color='k', label='valeur observée $x_j$')
plt.scatter([x0, x0], [norm.pdf(x0, 0, 1.2), norm.pdf(x0, 3, 1.5)], c=['tab:blue', 'tab:orange'], zorder=3)
plt.xlabel('valeur de la variable $x_j$'); plt.ylabel('densité $P(x_j\\mid c)$')
plt.title('Naive Bayes gaussien : comparer les vraisemblances'); plt.legend(); plt.show()
```

### Contextes d'utilisation

**Cas 1 – Finance :** Classer les transactions bancaires comme normales ou suspectes à partir de leurs caractéristiques.

**Cas 2 – Marketing :** Classer les avis des clients selon leur sentiment positif ou négatif.

**Nature des données et prérequis :** Données numériques ou textuelles transformées en variables exploitables. Pour Gaussian Naive Bayes, les variables explicatives sont supposées suivre une distribution gaussienne au sein de chaque classe.

**Cas inapproprié :** Il est moins adapté lorsque les variables sont fortement dépendantes les unes des autres, car il repose sur une hypothèse d'indépendance conditionnelle.

**Métriques :** Accuracy, précision, rappel, F1-score et matrice de confusion.

```python
from sklearn.tree import DecisionTreeClassifier
modele=DecisionTreeClassifier(max_depth=4,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

- **Nom, famille, objectif :** Arbre de décision (*Decision Tree Classifier*) — famille **classification**. Objectif : prédire une classe par une suite de règles de décision.
- **Intuition et étapes :** on partitionne récursivement l'espace pour obtenir des zones de plus en plus pures. Entrée : $X,y$ → à chaque nœud, choix de la coupure qui réduit le plus l'impureté → arrêt → prédiction = classe majoritaire de la feuille.
- **Formule :** impureté de Gini $G=1-\sum_{k=1}^{K}p_k^2$ ($K$ : nombre de classes, $p_k$ : proportion de la classe $k$ dans le nœud) ; alternative : entropie $-\sum_k p_k\log_2 p_k$.
- **Paramètres / hyperparamètres :** paramètres : variables et seuils de coupure ; hyperparamètres : profondeur maximale, taille minimale des feuilles ou des nœuds, critère d'impureté.
- **Avantages :** (1) règles lisibles et visualisables ; (2) pas de standardisation, gère des relations non linéaires.
- **Limites :** (1) instable (une petite variation des données change l'arbre) et surapprentissage si profond ; (2) frontières en escaliers parallèles aux axes.

### Q2. Schéma explicatif

**Comment le lire :** On lit l'arbre de haut en bas. Chaque nœud contient une question (test sur une variable) ; la branche de gauche correspond à « vrai », celle de droite à « faux ». Les feuilles (en bas) donnent la classe prédite ; la couleur indique la classe majoritaire et sa pureté, `gini` mesure l'impureté.

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.datasets import load_iris
iris = load_iris()
t = DecisionTreeClassifier(max_depth=2, random_state=42).fit(iris.data[:, 2:4], iris.target)
plt.figure(figsize=(9, 4.5))
plot_tree(t, feature_names=['longueur pétale', 'largeur pétale'], class_names=list(iris.target_names), filled=True, rounded=True)
plt.title("Arbre de décision (illustration sur le jeu Iris, profondeur 2)"); plt.show()
```

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

- **Nom, famille, objectif :** Machine à vecteurs de support (*Support Vector Machine, SVM*) — famille **classification**. Objectif : trouver la frontière séparant les classes avec la marge la plus large possible.
- **Intuition et étapes :** on cherche l'hyperplan équidistant des points les plus proches de chaque classe (vecteurs de support) ; un noyau permet de rendre la frontière non linéaire. Entrée : $X$ standardisée et $y$ → résolution d'un problème d'optimisation convexe → classe selon le côté de la frontière.
- **Formule :** $\min_{w,b,\xi}\ \frac12\|w\|^2+C\sum_i\xi_i$ sous $y_i(w^\top x_i+b)\ge 1-\xi_i$, $\xi_i\ge0$ ; noyau RBF : $K(x,x')=\exp(-\gamma\|x-x'\|^2)$ ($w,b$ : hyperplan, $\xi_i$ : variables de relâchement, $\gamma$ : portée du noyau).
- **Paramètres / hyperparamètres :** paramètres : $w,b$ (portés par les vecteurs de support) ; hyperparamètres : $C$, $\gamma$, type de noyau.
- **Avantages :** (1) efficace en dimension élevée ; (2) frontières non linéaires grâce aux noyaux.
- **Limites :** (1) coûteux sur de grands échantillons et sensible à l'échelle des variables ; (2) pas de probabilités directes et réglage de $C,\gamma$ délicat.

### Q2. Schéma explicatif

**Comment le lire :** La ligne continue est la frontière de décision ; les deux lignes pointillées délimitent la marge. Les points entourés (vecteurs de support) sont les seuls à définir la frontière : le SVM maximise la largeur de la marge.

```python
from sklearn.svm import SVC
rng = np.random.default_rng(42)
X = np.vstack([rng.normal([-2, -2], .8, (20, 2)), rng.normal([2, 2], .8, (20, 2))]); y = np.r_[np.zeros(20), np.ones(20)]
m = SVC(kernel='linear', C=1).fit(X, y); w = m.coef_[0]; b = m.intercept_[0]; xx = np.linspace(-5, 5, 50)
plt.figure(figsize=(5.5, 5)); plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolor='k')
for k, ls, lab in [(0, '-', 'frontière'), (1, '--', 'marge'), (-1, '--', None)]:
    plt.plot(xx, (-w[0]*xx - b + k)/w[1], 'k', ls=ls, label=lab)
plt.scatter(*m.support_vectors_.T, s=170, facecolors='none', edgecolors='g', lw=2, label='vecteurs de support')
plt.xlim(-5, 5); plt.ylim(-5, 5); plt.title('SVM linéaire : frontière à marge maximale'); plt.legend(); plt.show()
```

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import AdaBoostClassifier
modele=AdaBoostClassifier(n_estimators=70,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

- **Nom, famille, objectif :** AdaBoost (*Adaptive Boosting*) — famille **classification** (ensemble séquentiel). Objectif : combiner de nombreux classifieurs faibles (souvent des souches d'arbre) en un classifieur fort.
- **Intuition et étapes :** après chaque classifieur faible, on augmente le poids des observations mal classées pour que le suivant se concentre dessus. Entrée : $X,y$, poids uniformes → entraînement pondéré → calcul de l'erreur et du poids du classifieur → mise à jour des poids → vote pondéré.
- **Formule :** $F(x)=\text{signe}\big(\sum_{m=1}^{M}\alpha_m h_m(x)\big)$ avec $\alpha_m=\frac12\ln\frac{1-\varepsilon_m}{\varepsilon_m}$ (cas binaire ; $h_m$ : classifieur faible, $\varepsilon_m$ : son erreur pondérée, $\alpha_m$ : son poids dans le vote).
- **Paramètres / hyperparamètres :** paramètres : classifieurs $h_m$ et poids $\alpha_m$ ; hyperparamètres : nombre d'estimateurs $M$, taux d'apprentissage, type de classifieur faible.
- **Avantages :** (1) simple et peu de réglages ; (2) améliore nettement des modèles faibles.
- **Limites :** (1) sensible au bruit et aux valeurs aberrantes (les erreurs sont sur-pondérées) ; (2) entraînement séquentiel non parallélisable.

### Q2. Schéma explicatif

**Comment le lire :** Chaque panneau est un tour de boosting. La taille d'un point indique son poids : les points mal classés par la souche du tour (cerclés en noir) deviennent plus gros au tour suivant. Le trait vert est la coupure de la souche.

```python
from sklearn.tree import DecisionTreeClassifier
rng = np.random.default_rng(42)
X = rng.normal(size=(40, 2)); y = (X[:, 0] + X[:, 1] > 0).astype(int); w = np.ones(40)/40
fig, axs = plt.subplots(1, 3, figsize=(12, 3.8))
for r, ax in enumerate(axs):
    s = DecisionTreeClassifier(max_depth=1, random_state=42).fit(X, y, sample_weight=w); p = s.predict(X)
    err = max(w[p != y].sum(), 1e-6); a = .5*np.log((1-err)/err)
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', s=w*3000, edgecolors=np.where(p != y, 'k', 'none'), linewidths=1.8)
    f, th = s.tree_.feature[0], s.tree_.threshold[0]
    (ax.axvline if f == 0 else ax.axhline)(th, color='g', lw=2)
    ax.set_title(f'tour {r+1} : erreur pondérée = {err:.2f}')
    w = w*np.exp(np.where(p != y, a, -a)); w /= w.sum()
fig.suptitle('AdaBoost : les erreurs reçoivent plus de poids au tour suivant'); plt.tight_layout(); plt.show()
```

## 16. XGBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from xgboost import XGBClassifier
modele=XGBClassifier(n_estimators=80,max_depth=3,random_state=GRAINE,eval_metric='logloss').fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Présentation

- **Nom, famille, objectif :** XGBoost (*eXtreme Gradient Boosting*) — famille **classification** (ensemble séquentiel), disponible aussi en régression. Objectif : classer avec un boosting d'arbres optimisé et régularisé.
- **Intuition et étapes :** boosting de gradient qui utilise une approximation de second ordre de la perte et une pénalité sur la complexité des arbres. Entrée : $X,y$ → ajout séquentiel d'arbres régularisés → somme pondérée des sorties → probabilité de classe.
- **Formule :** $\mathcal L=\sum_{i}\ell(y_i,\hat y_i)+\sum_{k}\Omega(f_k)$ avec $\Omega(f)=\gamma T+\frac12\lambda\|w\|^2$ ($\ell$ : perte, $f_k$ : arbre $k$, $T$ : nombre de feuilles, $w$ : valeurs des feuilles, $\gamma,\lambda$ : poids de régularisation).
- **Paramètres / hyperparamètres :** paramètres : arbres et valeurs de feuilles ; hyperparamètres : nombre d'arbres, profondeur maximale, taux d'apprentissage, sous-échantillonnage, $\gamma$, $\lambda$.
- **Avantages :** (1) très performant et rapide grâce à des optimisations d'implémentation ; (2) régularisation intégrée, gère les valeurs manquantes.
- **Limites :** (1) nombreux hyperparamètres à régler ; (2) peu interprétable et risque de surapprentissage si mal réglé.

```python
from sklearn.cluster import KMeans
modele=KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(X_clu)
print('Silhouette:',silhouette_score(X_clu,modele.labels_))
plt.scatter(X_clu[:,0],X_clu[:,1],c=modele.labels_,s=12);plt.title('K-means');plt.show()
```

### Présentation

- **Nom, famille, objectif :** K-means (*K-Means*, K-moyennes) — famille **clustering** (non supervisé). Objectif : regrouper les observations en $K$ groupes compacts.
- **Intuition et étapes :** chaque groupe est représenté par son centre (centroïde). Entrée : $X$ sans étiquettes → initialisation de $K$ centroïdes → affectation de chaque point au centroïde le plus proche → recalcul des centroïdes → répétition jusqu'à stabilité → sortie : une étiquette de groupe par point.
- **Formule :** on minimise l'inertie $J=\sum_{k=1}^{K}\sum_{x_i\in C_k}\|x_i-\mu_k\|^2$ ($C_k$ : groupe $k$, $\mu_k$ : son centroïde).
- **Paramètres / hyperparamètres :** paramètres : positions des centroïdes ; hyperparamètres : nombre de groupes $K$, nombre d'initialisations, méthode d'initialisation.
- **Avantages :** (1) simple, rapide, adapté aux grands jeux de données ; (2) résultat facile à interpréter via les centroïdes.
- **Limites :** (1) $K$ doit être fixé à l'avance et le résultat dépend de l'initialisation ; (2) suppose des groupes plutôt sphériques de taille comparable et est sensible aux valeurs extrêmes.

### Q2. Schéma explicatif

**Comment le lire :** Chaque panneau montre une itération. Les couleurs donnent le groupe de chaque point ; les croix noires sont les centroïdes. On voit les centroïdes se déplacer vers le centre de leur groupe, puis la partition se stabiliser.

```python
from sklearn.datasets import make_blobs
X, _ = make_blobs(n_samples=150, centers=3, cluster_std=.9, random_state=42)
rng = np.random.default_rng(42); C0 = X[rng.choice(len(X), 3, replace=False)]
def affecter(X, C): return np.argmin(((X[:, None] - C[None])**2).sum(2), axis=1)
fig, axs = plt.subplots(1, 3, figsize=(12, 3.8)); C = C0.copy()
for it, ax in zip([1, 2, 6], axs):
    C = C0.copy()
    for _ in range(it - 1):
        lab = affecter(X, C); C = np.array([X[lab == k].mean(0) if (lab == k).any() else C[k] for k in range(3)])
    lab = affecter(X, C)
    ax.scatter(X[:, 0], X[:, 1], c=lab, cmap='viridis', s=12); ax.scatter(*C.T, marker='X', s=160, c='k')
    ax.set_title(f'itération {it}')
fig.suptitle('K-means : affectation puis recalcul des centroïdes (X noir = centroïde)'); plt.tight_layout(); plt.show()
```

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
```

### Présentation

- **Nom, famille, objectif :** Classification hiérarchique ascendante (*Agglomerative Hierarchical Clustering*) — famille **clustering**. Objectif : construire une hiérarchie de regroupements emboîtés, représentée par un dendrogramme.
- **Intuition et étapes :** chaque point est d'abord son propre groupe, puis on fusionne à chaque étape les deux groupes les plus proches. Entrée : $X$ (ou matrice de distances) → fusions successives selon un critère de lien → dendrogramme → coupe à un niveau donné pour obtenir $K$ groupes.
- **Formule :** critère de Ward : coût de fusion $\Delta(A,B)=\dfrac{n_An_B}{n_A+n_B}\|\mu_A-\mu_B\|^2$ ($n_A,n_B$ : effectifs, $\mu_A,\mu_B$ : centres des groupes $A$ et $B$).
- **Paramètres / hyperparamètres :** pas de paramètres appris au sens strict ; hyperparamètres : nombre de groupes (niveau de coupe), type de lien (Ward, complet, moyen, simple), distance.
- **Avantages :** (1) le dendrogramme montre la structure à plusieurs échelles ; (2) pas besoin de fixer $K$ à l'avance pour construire la hiérarchie.
- **Limites :** (1) coût en temps et mémoire d'au moins $O(n^2)$, peu adapté aux très grands jeux de données ; (2) une fusion est irréversible et sensible au choix du lien.

### Q2. Schéma explicatif

**Comment le lire :** Les feuilles (en bas) sont les observations. Quand on monte, des groupes fusionnent ; la hauteur d'une fusion indique la distance entre les groupes fusionnés. La ligne rouge horizontale coupe l'arbre : le nombre de branches qu'elle traverse est le nombre de groupes retenus (ici 3).

```python
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.datasets import make_blobs
X, _ = make_blobs(n_samples=24, centers=3, cluster_std=.8, random_state=42)
Z = linkage(X, method='ward'); h = (Z[-3, 2] + Z[-2, 2]) / 2
plt.figure(figsize=(8, 4)); dendrogram(Z, color_threshold=h); plt.axhline(h, color='r', ls='--', label='coupe → 3 groupes')
plt.xlabel('observations'); plt.ylabel('distance de fusion (Ward)'); plt.title('Classification hiérarchique : dendrogramme'); plt.legend(); plt.show()
```

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

### Présentation

- **Nom, famille, objectif :** DBSCAN (*Density-Based Spatial Clustering of Applications with Noise*) — famille **clustering** (basé sur la densité), utilisable aussi pour repérer du bruit. Objectif : trouver des groupes de forme quelconque et isoler les points isolés.
- **Intuition et étapes :** un groupe est une zone dense séparée de zones peu denses. Entrée : $X$ → pour chaque point, comptage des voisins dans un rayon $\varepsilon$ → points « cœur » reliés entre eux pour former les groupes → points non rattachés étiquetés bruit (−1).
- **Formule :** voisinage $N_\varepsilon(x)=\{x' : \|x-x'\|\le\varepsilon\}$ ; $x$ est un point cœur si $|N_\varepsilon(x)|\ge \text{min\_samples}$ ($\varepsilon$ : rayon, min_samples : effectif minimal).
- **Paramètres / hyperparamètres :** pas de paramètres appris ; hyperparamètres : $\varepsilon$ (`eps`), `min_samples`, distance.
- **Avantages :** (1) détecte des formes arbitraires sans fixer le nombre de groupes ; (2) identifie explicitement le bruit.
- **Limites :** (1) très sensible au choix de $\varepsilon$ et de min_samples ; (2) difficulté avec des densités variables ou en dimension élevée.

### Q2. Schéma explicatif

**Comment le lire :** Les grands points colorés sont des points « cœur » (au moins `min_samples` voisins dans le cercle de rayon $\varepsilon$), les petits points colorés sont en bordure d'un groupe, les croix noires sont du bruit. Le cercle pointillé montre le voisinage d'un point.

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
X, _ = make_moons(n_samples=200, noise=.07, random_state=42)
db = DBSCAN(eps=.2, min_samples=5).fit(X); lab = db.labels_
coeur = np.zeros(len(X), bool); coeur[db.core_sample_indices_] = True
fig, ax = plt.subplots(figsize=(6, 4.5))
ax.scatter(*X[coeur].T, c=lab[coeur], cmap='coolwarm', s=45, label='point cœur')
ax.scatter(*X[~coeur & (lab >= 0)].T, c=lab[~coeur & (lab >= 0)], cmap='coolwarm', s=12, label='point de bordure')
ax.scatter(*X[lab == -1].T, marker='x', c='k', label='bruit')
ax.add_patch(Circle(X[db.core_sample_indices_[0]], .2, fill=False, ls='--')); ax.set_aspect('equal')
ax.set_title(r'DBSCAN ($\varepsilon$ = 0,2 ; min_samples = 5)'); ax.legend(); plt.show()
```

### Présentation

- **Nom, famille, objectif :** Mélanges gaussiens (*Gaussian Mixture Models, GMM*) — famille **clustering** (probabiliste). Objectif : modéliser les données comme un mélange de lois normales et affecter chaque point à une composante avec une probabilité.
- **Intuition et étapes :** chaque groupe est une gaussienne (centre, forme, poids). Entrée : $X$ → initialisation → algorithme EM : étape E (probabilité d'appartenance de chaque point à chaque composante), étape M (mise à jour des poids, moyennes, covariances) → répétition jusqu'à convergence → affectation douce ou dure.
- **Formule :** $p(x)=\sum_{k=1}^{K}\pi_k\,\mathcal N(x\mid\mu_k,\Sigma_k)$ ($\pi_k$ : poids de la composante $k$ avec $\sum_k\pi_k=1$, $\mu_k$ : moyenne, $\Sigma_k$ : matrice de covariance).
- **Paramètres / hyperparamètres :** paramètres : $\pi_k,\mu_k,\Sigma_k$ ; hyperparamètres : nombre de composantes $K$, type de covariance, initialisation.
- **Avantages :** (1) affectation probabiliste (incertitude) ; (2) groupes elliptiques de tailles différentes.
- **Limites :** (1) peut converger vers un optimum local et exige de choisir $K$ (par exemple avec BIC/AIC) ; (2) suppose des composantes gaussiennes.

### Q2. Schéma explicatif

**Comment le lire :** Chaque ellipse (contours à 1 et 2 écarts-types) représente une composante gaussienne : son centre, sa forme et son orientation. La couleur d'un point indique la certitude de son affectation : les points proches de deux ellipses ont une probabilité partagée.

```python
from sklearn.mixture import GaussianMixture
from sklearn.datasets import make_blobs
X, _ = make_blobs(n_samples=300, centers=3, cluster_std=[1, 1.8, .7], random_state=42)
g = GaussianMixture(3, random_state=42).fit(X)
fig, ax = plt.subplots(figsize=(6, 4.5)); sc = ax.scatter(X[:, 0], X[:, 1], c=g.predict_proba(X).max(1), cmap='viridis', s=12)
for mu, cov in zip(g.means_, g.covariances_):
    vals, vecs = np.linalg.eigh(cov); ang = np.degrees(np.arctan2(vecs[1, 1], vecs[0, 1]))
    for k in (1, 2): ax.add_patch(Ellipse(mu, 2*k*np.sqrt(vals[1]), 2*k*np.sqrt(vals[0]), angle=ang, fill=False, color='r'))
plt.colorbar(sc, label="probabilité maximale d'appartenance"); ax.set_title('Mélange gaussien : 3 composantes'); plt.show()
```

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.decomposition import PCA
X_vin=StandardScaler().fit_transform(load_wine().data)
modele=PCA(n_components=2).fit(X_vin)
projection=modele.transform(X_vin)
print('Variance expliquée:',modele.explained_variance_ratio_)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.show()
```

### Présentation

- **Nom, famille, objectif :** Analyse en composantes principales (ACP, *Principal Component Analysis, PCA*) — famille **réduction de dimension** (linéaire, non supervisée). Objectif : résumer les données par quelques axes conservant le maximum de variance.
- **Intuition et étapes :** on cherche de nouveaux axes orthogonaux le long desquels les données sont les plus étalées. Entrée : $X$ centré (et standardisé) → calcul de la matrice de covariance et de ses vecteurs propres → projection sur les $q$ premiers axes → sortie : coordonnées réduites.
- **Formule :** $\Sigma v_k=\lambda_k v_k$ ; part de variance expliquée par l'axe $k$ : $\lambda_k/\sum_{j}\lambda_j$ ($\Sigma$ : matrice de covariance, $v_k$ : axe principal, $\lambda_k$ : variance portée par cet axe).
- **Paramètres / hyperparamètres :** paramètres : axes $v_k$ et valeurs propres $\lambda_k$ ; hyperparamètre : nombre de composantes conservées $q$ (ou seuil de variance).
- **Avantages :** (1) rapide, déterministe et interprétable via la variance expliquée ; (2) réduit le bruit et la redondance.
- **Limites :** (1) ne capte que des structures linéaires ; (2) sensible à l'échelle des variables (standardisation nécessaire) et axes parfois difficiles à nommer.

### Q2. Schéma explicatif

**Comment le lire :** Le nuage de points est allongé selon une direction. La flèche rouge (CP1) pointe dans la direction de plus grande variance ; la flèche verte (CP2) est perpendiculaire et porte la variance restante. Les pourcentages indiquent la part de variance expliquée par chaque axe.

```python
from sklearn.decomposition import PCA
rng = np.random.default_rng(42); X = rng.multivariate_normal([0, 0], [[3, 2], [2, 2]], 200); p = PCA(2).fit(X)
plt.figure(figsize=(5.5, 5)); plt.scatter(X[:, 0], X[:, 1], s=10, alpha=.5, label='observations')
for i, (v, c) in enumerate(zip(p.components_, ['r', 'g'])):
    f = 2.5*np.sqrt(p.explained_variance_[i])
    plt.annotate('', xy=v*f, xytext=(0, 0), arrowprops=dict(arrowstyle='-|>', color=c, lw=2.5))
    plt.text(*(v*f*1.12), f'CP{i+1} : {100*p.explained_variance_ratio_[i]:.0f} %', color=c)
plt.axis('equal'); plt.xlabel('variable 1'); plt.ylabel('variable 2'); plt.title('ACP : axes de variance maximale'); plt.legend(); plt.show()
```

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import IsolationForest
X_anomal=np.vstack([X_clu,np.array([[9,9],[-8,-8]])])
modele=IsolationForest(contamination=.02,random_state=GRAINE).fit(X_anomal)
print('Observations atypiques:',np.where(modele.predict(X_anomal)==-1)[0])
```

### Présentation

- **Nom, famille, objectif :** Forêt d'isolement (*Isolation Forest*) — famille **détection d'anomalies** (non supervisé). Objectif : repérer les observations atypiques.
- **Intuition et étapes :** une anomalie, rare et éloignée, est isolée avec très peu de coupures aléatoires. Entrée : $X$ → construction de nombreux arbres de coupures aléatoires (variable et seuil tirés au hasard) → profondeur moyenne d'isolement de chaque point → score d'anomalie → seuil selon la contamination attendue.
- **Formule :** $s(x,n)=2^{-E[h(x)]/c(n)}$ ($h(x)$ : profondeur à laquelle $x$ est isolé, $E[h(x)]$ : sa moyenne sur les arbres, $c(n)$ : profondeur moyenne de normalisation pour $n$ points ; $s$ proche de 1 = anomalie probable).
- **Paramètres / hyperparamètres :** paramètres : arbres d'isolement ; hyperparamètres : nombre d'arbres, taille du sous-échantillon, taux de contamination.
- **Avantages :** (1) rapide, passe à l'échelle et ne demande pas d'étiquettes ; (2) fonctionne en dimension relativement élevée.
- **Limites :** (1) la contamination doit être fixée ou estimée ; (2) moins adapté aux anomalies locales ou contextuelles, et score difficile à interpréter.

### Q2. Schéma explicatif

**Comment le lire :** La couleur donne le score d'anomalie calculé par la forêt : rouge = zone où un point est isolé très vite (très atypique), bleu = zone dense (normale). Les points isolés situés dans les zones rouges (étoiles) sont signalés comme anomalies.

```python
from sklearn.ensemble import IsolationForest
rng = np.random.default_rng(42); X = np.vstack([rng.normal(0, 1, (150, 2)), [[5, 5], [-5, 4.5]]])
m = IsolationForest(contamination=.02, random_state=42).fit(X)
xx, yy = np.meshgrid(np.linspace(-7, 7, 200), np.linspace(-7, 7, 200))
z = m.decision_function(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape); pred = m.predict(X)
plt.figure(figsize=(6, 5)); plt.contourf(xx, yy, z, levels=15, cmap='RdYlBu'); plt.colorbar(label='score (bas = plus atypique)')
plt.scatter(*X[pred == 1].T, c='w', edgecolor='k', s=18, label='normal'); plt.scatter(*X[pred == -1].T, c='k', marker='*', s=140, label='anomalie détectée')
plt.title("Isolation Forest : carte des scores d'anomalie"); plt.legend(); plt.show()
```

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
```

### Présentation

- **Nom, famille, objectif :** Bandits manchots (*Multi-Armed Bandits*) — famille **renforcement** (cas à un seul état). Objectif : choisir à chaque tour parmi plusieurs actions (bras) pour maximiser la récompense cumulée, sans connaître leurs gains moyens.
- **Intuition et étapes :** il faut arbitrer entre explorer (tester les bras) et exploiter (jouer le meilleur bras estimé). Boucle : choisir un bras → observer une récompense → mettre à jour l'estimation de ce bras. Dans l'exemple, les bras sont joués à tour de rôle (exploration pure), ce qui sert de référence.
- **Formule :** estimation $\hat\mu_a=\frac{1}{N_a}\sum_{t:a_t=a}r_t$ ; regret $R_T=T\mu^*-\sum_{t=1}^{T}r_t$ ($a$ : bras, $N_a$ : nombre de tirages du bras $a$, $r_t$ : récompense au temps $t$, $\mu^*$ : gain moyen du meilleur bras, $T$ : nombre de tours).
- **Paramètres / hyperparamètres :** paramètres estimés : valeurs $\hat\mu_a$ ; hyperparamètres : règle de sélection et ses réglages, horizon $T$.
- **Avantages :** (1) modèle simple pour décider sous incertitude ; (2) apprentissage en ligne, sans jeu de données préalable.
- **Limites :** (1) la stratégie uniforme de l'exemple gaspille des essais sur les mauvais bras ; (2) ignore le contexte et l'effet des actions sur l'état futur.

### Q2. Schéma explicatif

**Comment le lire :** On suit la boucle : l'agent choisit un bras, la machine renvoie une récompense (0 ou 1), l'agent met à jour son estimation du gain moyen de ce bras, puis recommence (flèche rouge). Le défi est de choisir entre explorer et exploiter.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Présentation

- **Nom, famille, objectif :** Epsilon-greedy (*ε-greedy*, stratégie epsilon-glouton) — famille **renforcement** (bandits). Objectif : équilibrer exploration et exploitation pour maximiser la récompense cumulée.
- **Intuition et étapes :** la plupart du temps on joue le bras au meilleur gain estimé, mais avec une petite probabilité $\varepsilon$ on en joue un au hasard. Boucle : tirage aléatoire → choix du bras (aléatoire ou meilleur) → récompense → mise à jour de la moyenne du bras.
- **Formule :** $a_t=\begin{cases}\text{bras aléatoire}&\text{avec probabilité }\varepsilon\\ \arg\max_a\hat\mu_a&\text{avec probabilité }1-\varepsilon\end{cases}$ ($\hat\mu_a$ : gain moyen estimé du bras $a$).
- **Paramètres / hyperparamètres :** paramètres : estimations $\hat\mu_a$ ; hyperparamètres : $\varepsilon$ (parfois décroissant), horizon.
- **Avantages :** (1) très simple à implémenter ; (2) garantit de continuer à explorer tous les bras.
- **Limites :** (1) explore au hasard, même quand les bras sont déjà bien connus (regret linéaire si $\varepsilon$ est constant) ; (2) sensible au choix de $\varepsilon$.

### Q2. Schéma explicatif

**Comment le lire :** On lit de haut en bas. À chaque tour on tire un nombre $u$ entre 0 et 1 : s'il est inférieur à $\varepsilon$, on explore (bras au hasard), sinon on exploite (meilleur bras estimé). Dans les deux cas, on met ensuite à jour l'estimation du bras joué.

```python
fig, ax = plt.subplots(figsize=(8, 5)); ax.set_xlim(0, 10); ax.set_ylim(0, 6); ax.axis('off')
boite(ax, 5, 5.2, r'Tirer $u\in[0,1]$ au hasard', l=3.6)
boite(ax, 2.5, 3.0, 'EXPLORER\nchoisir un bras\nau hasard', l=3.2, h=1.3, couleur='#fde9d9')
boite(ax, 7.5, 3.0, 'EXPLOITER\nchoisir le bras\n' + r'$\arg\max_a\hat{\mu}_a$', l=3.2, h=1.3, couleur='#e2f0d9')
boite(ax, 5, 0.8, r'Observer $r_t$, mettre à jour $\hat{\mu}_a$', l=4.2)
fleche(ax, (4, 4.7), (2.8, 3.7), r'$u<\varepsilon$', dy=0.3); fleche(ax, (6, 4.7), (7.2, 3.7), r'$u\geq\varepsilon$', dy=0.3)
fleche(ax, (2.8, 2.3), (4.2, 1.3)); fleche(ax, (7.2, 2.3), (5.8, 1.3))
ax.set_title('Epsilon-greedy : règle de décision'); plt.show()
```

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### Présentation

- **Nom, famille, objectif :** Q-learning (*Q-Learning*) — famille **renforcement** (sans modèle, hors politique). Objectif : apprendre une politique optimale dans un environnement à états discrets en estimant la valeur de chaque couple état-action.
- **Intuition et étapes :** l'agent essaie des actions, observe la récompense et l'état suivant, et améliore une table $Q$. Boucle : état → action (ε-greedy) → récompense et nouvel état → mise à jour de $Q$ → répétition sur de nombreux épisodes ; la politique finale est l'action de $Q$ maximal dans chaque état.
- **Formule :** $Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\big]$ ($s,a$ : état et action, $r$ : récompense, $s'$ : état suivant, $\alpha$ : taux d'apprentissage, $\gamma$ : facteur d'actualisation).
- **Paramètres / hyperparamètres :** paramètres : la table $Q$ ; hyperparamètres : $\alpha$, $\gamma$, $\varepsilon$ (et sa décroissance), nombre d'épisodes.
- **Avantages :** (1) simple et converge vers la politique optimale sous conditions ; (2) apprend sans connaître le modèle de l'environnement.
- **Limites :** (1) la table devient impraticable quand les états sont nombreux ou continus ; (2) sensible aux hyperparamètres et peut être lent à converger.

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### Présentation

- **Nom, famille, objectif :** SARSA (*State-Action-Reward-State-Action*) — famille **renforcement** (sans modèle, sur politique). Objectif : apprendre les valeurs $Q(s,a)$ de la politique effectivement suivie.
- **Intuition et étapes :** proche du Q-learning, mais la mise à jour utilise l'action réellement choisie ensuite, y compris les actions exploratoires. Boucle : état $s$, action $a$ → récompense $r$ et état $s'$ → choix de $a'$ par la même politique → mise à jour de $Q(s,a)$.
- **Formule :** $Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma Q(s',a')-Q(s,a)\big]$ ($a'$ : action réellement choisie dans l'état $s'$ ; autres symboles comme pour le Q-learning).
- **Paramètres / hyperparamètres :** paramètres : la table $Q$ ; hyperparamètres : $\alpha$, $\gamma$, $\varepsilon$ (et sa décroissance), nombre d'épisodes.
- **Avantages :** (1) tient compte de l'exploration, donc apprend des trajectoires plus prudentes dans les environnements risqués ; (2) simple et peu coûteux.
- **Limites :** (1) converge vers la politique optimale seulement si $\varepsilon$ décroît vers 0 ; (2) même limite de passage à l'échelle que le Q-learning tabulaire.

### Q2. Schéma explicatif

**Comment le lire :** La chaîne du haut est une transition : état $s$, action $a$, puis récompense $r$ et état $s'$, puis l'action suivante $a'$. SARSA utilise l'action $a'$ réellement choisie (politique suivie) dans sa cible ; le Q-learning utilise le meilleur $Q(s',\cdot)$, même si cette action n'est pas celle qui sera jouée.

```python
fig, ax = plt.subplots(figsize=(10, 4.2)); ax.set_xlim(0, 11); ax.set_ylim(0, 5); ax.axis('off')
noms = ['état $s$', 'action $a$', "récompense $r$,\nétat $s'$", "action $a'$\n(réellement choisie)"]
for i, t in enumerate(noms):
    boite(ax, 1.4 + 2.7*i, 4, t, l=2.0, h=1.2)
    if i < 3: fleche(ax, (2.5 + 2.7*i, 4), (3.0 + 2.7*i, 4))
boite(ax, 8.6, 1.3, "SARSA (sur politique) :\ncible $= r+\\gamma\\,Q(s',a')$", l=4.2, h=1.3, couleur='#e2f0d9')
boite(ax, 3.2, 1.3, "Q-learning (hors politique) :\ncible $= r+\\gamma\\max_{a'}Q(s',a')$", l=4.8, h=1.3, couleur='#fde9d9')
fleche(ax, (6.8, 3.4), (8.2, 1.95)); fleche(ax, (5.4, 3.4), (3.6, 1.95))
ax.set_title('SARSA et Q-learning : où se différencie la cible'); plt.show()
```

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
# Calcul de cible DQN, sans entraîner un réseau neuronal complet
recompenses=np.array([0.,1.,-1.]);termine=np.array([False,True,False]);meilleur_Q_suivant=np.array([2.,3.,1.]);gamma=.95
cibles=recompenses+gamma*(~termine)*meilleur_Q_suivant
print('Cibles de Bellman:',cibles)
# Un DQN complet ajoute réseau en ligne, réseau cible et mémoire de rejeu.
```

### Présentation

- **Nom, famille, objectif :** Réseau Q profond (*Deep Q-Network, DQN*) — famille **renforcement** (approximation de fonction). Objectif : apprendre $Q(s,a)$ avec un réseau de neurones lorsque l'espace d'états est trop grand pour une table. L'exemple du notebook ne calcule que la cible de Bellman (version conceptuelle).
- **Intuition et étapes :** un réseau remplace la table $Q$. Boucle : l'agent interagit et stocke ses transitions dans une mémoire de rejeu → tirage de lots de transitions → calcul de la cible à l'aide d'un réseau cible figé → descente de gradient sur le réseau en ligne → mise à jour périodique du réseau cible.
- **Formule :** cible $y=r+\gamma(1-d)\max_{a'}Q_{\theta^-}(s',a')$ et perte $L(\theta)=\mathbb E\big[(y-Q_\theta(s,a))^2\big]$ ($d\in\{0,1\}$ : fin d'épisode, $\theta$ : poids du réseau en ligne, $\theta^-$ : poids du réseau cible).
- **Paramètres / hyperparamètres :** paramètres : poids $\theta$ du réseau ; hyperparamètres : $\gamma$, taux d'apprentissage, taille de la mémoire et des lots, fréquence de mise à jour du réseau cible, $\varepsilon$.
- **Avantages :** (1) gère des états de grande dimension (images, capteurs) ; (2) généralise entre états proches.
- **Limites :** (1) entraînement instable, gourmand en données et en calcul ; (2) résultats sensibles aux hyperparamètres et à la graine.

### Q2. Schéma explicatif

**Comment le lire :** On suit les flèches : l'agent agit dans l'environnement ; chaque transition est stockée dans la mémoire de rejeu. Des mini-lots tirés au hasard alimentent le réseau en ligne (valeur actuelle) et le réseau cible (cible de Bellman). La perte compare les deux, et le réseau cible est recopié périodiquement.

```python
fig, ax = plt.subplots(figsize=(11, 5)); ax.set_xlim(0, 12.5); ax.set_ylim(0, 6); ax.axis('off')
boite(ax, 1.3, 3, 'Environnement', l=2.2, h=1.1)
boite(ax, 4.9, 3, 'Mémoire de rejeu\n$(s,a,r,s\')$', l=2.6, h=1.2, couleur='#e2f0d9')
boite(ax, 8.0, 4.5, 'Réseau en ligne\n$Q_\\theta(s,a)$', l=2.6, h=1.2)
boite(ax, 8.0, 1.5, 'Réseau cible\n$Q_{\\theta^-}$', l=2.6, h=1.2, couleur='#fde9d9')
boite(ax, 11.2, 3, 'Perte\n$(y-Q_\\theta)^2$', l=2.0, h=1.2, couleur='#f8cbad')
fleche(ax, (2.45, 3), (3.55, 3), "$(s,a,r,s')$", dy=.3)
fleche(ax, (6.25, 3.4), (6.7, 4.3), 'mini-lot', dy=.35); fleche(ax, (6.25, 2.6), (6.7, 1.7), 'mini-lot', dy=-.6)
fleche(ax, (9.3, 4.3), (10.2, 3.4), '$Q_\\theta(s,a)$', dy=.3); fleche(ax, (9.3, 1.7), (10.2, 2.6), 'cible $y$', dy=-.55)
fleche(ax, (8.0, 3.9), (8.0, 2.1), 'copie tous\nles C pas', style='--', dy=-.1)
fleche(ax, (7.0, 5.1), (1.3, 3.6), "action $\\varepsilon$-greedy", rad=.25, dy=.9)
ax.set_title('DQN : architecture (mémoire de rejeu + réseau cible)'); plt.show()
```
